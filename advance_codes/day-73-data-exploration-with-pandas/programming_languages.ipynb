### Método 3: Crear datos de ejemplo

### Estadísticas descriptivas

## 4️⃣ Selección de Datos

### Seleccionar columnas

````python
# Seleccionar una sola columna (devuelve una Series)
nombres = df['nombre']
print("👤 Columna 'nombre':")
print(nombres)
print(f"\nTipo: {type(nombres)}")
````

````python
# Seleccionar múltiples columnas (devuelve un DataFrame)
# Nota: usar doble corchete [[ ]]
subset = df[['nombre', 'edad', 'salario']]
print("📋 Columnas seleccionadas:")
print(subset)
print(f"\nTipo: {type(subset)}")
````

### Seleccionar filas

````python
# .loc[] selecciona por etiqueta/índice
# Sintaxis: df.loc[filas, columnas]

# Seleccionar la primera fila
print("📍 Primera fila (índice 0):")
print(df.loc[0])

print("\n" + "="*50 + "\n")

# Seleccionar filas del índice 0 al 2
print("📍 Filas 0 a 2:")
print(df.loc[0:2])
````

````python
# .iloc[] selecciona por posición numérica
# Similar a usar índices en listas de Python

# Seleccionar la primera fila
print("📍 Primera fila (posición 0):")
print(df.iloc[0])

print("\n" + "="*50 + "\n")

# Seleccionar las primeras 3 filas
print("📍 Primeras 3 filas:")
print(df.iloc[0:3])
````

````python
# Combinación: seleccionar filas y columnas específicas
# df.loc[filas, columnas]

# Seleccionar nombre y edad de las primeras 3 personas
print("📋 Nombres y edades (primeras 3 filas):")
print(df.loc[0:2, ['nombre', 'edad']])
````

## 5️⃣ Filtrado de Datos

### Filtros simples

````python
# Filtrar personas mayores de 30 años
# df['columna'] > valor crea una máscara booleana (True/False)
# df[mascara] devuelve solo las filas donde la máscara es True

mayores_30 = df[df['edad'] > 30]
print("👥 Personas mayores de 30 años:")
print(mayores_30)
````

````python
# Filtrar personas de Madrid
# Para comparar texto, usar ==

madrid = df[df['ciudad'] == 'Madrid']
print("🏙️ Personas de Madrid:")
print(madrid)
````

### Filtros múltiples

````python
# Operador AND: usar &
# Importante: cada condición debe estar entre paréntesis

filtro_and = df[(df['edad'] > 30) & (df['ciudad'] == 'Madrid')]
print("🔍 Mayores de 30 años Y de Madrid:")
print(filtro_and)
````

````python
# Operador OR: usar |
# También requiere paréntesis en cada condición

filtro_or = df[(df['edad'] > 35) | (df['salario'] > 50000)]
print("🔍 Mayores de 35 años O salario mayor a 50000:")
print(filtro_or)
````

````python
# Operador NOT: usar ~
# Invierte la condición

no_madrid = df[~(df['ciudad'] == 'Madrid')]
print("🔍 Personas que NO son de Madrid:")
print(no_madrid)
````

### Filtros con .isin()

````python
# .isin() verifica si el valor está en una lista
# Útil para filtrar por múltiples valores

ciudades_principales = df[df['ciudad'].isin(['Madrid', 'Barcelona'])]
print("🏙️ Personas de Madrid o Barcelona:")
print(ciudades_principales)
````

## 6️⃣ Ordenar Datos

````python
# .sort_values() ordena por una o más columnas
# ascending=True: orden ascendente (menor a mayor)
# ascending=False: orden descendente (mayor a menor)

# Ordenar por edad (ascendente)
ordenado_edad = df.sort_values('edad')
print("📈 Ordenado por edad (menor a mayor):")
print(ordenado_edad)
````

````python
# Ordenar por salario (descendente)
ordenado_salario = df.sort_values('salario', ascending=False)
print("💰 Ordenado por salario (mayor a menor):")
print(ordenado_salario)
````

````python
# Ordenar por múltiples columnas
# Primero por ciudad (A-Z), luego por edad (menor a mayor)

ordenado_multiple = df.sort_values(['ciudad', 'edad'])
print("📊 Ordenado por ciudad y edad:")
print(ordenado_multiple)
````

## 7️⃣ Agregar y Modificar Columnas

````python
# Agregar una nueva columna
# Sintaxis: df['nueva_columna'] = valores

# Calcular salario anual (salario * 12)
df['salario_anual'] = df['salario'] * 12
print("💼 DataFrame con salario anual:")
print(df)
````

````python
# Crear columna basada en condiciones
# np.where(condición, valor_si_true, valor_si_false)

df['categoria_edad'] = np.where(df['edad'] > 35, 'Senior', 'Junior')
print("👥 DataFrame con categoría de edad:")
print(df[['nombre', 'edad', 'categoria_edad']])
````

````python
# Modificar valores de una columna existente
# Aumentar todos los salarios en 10%

df['salario'] = df['salario'] * 1.10
print("💰 Salarios aumentados 10%:")
print(df[['nombre', 'salario']])
````

## 8️⃣ Eliminar Datos

````python
# .drop() elimina filas o columnas
# axis=1: eliminar columnas
# axis=0: eliminar filas
# inplace=False: crear copia (no modifica el original)
# inplace=True: modificar el DataFrame original

# Eliminar columna 'salario_anual' (crear copia)
df_sin_anual = df.drop('salario_anual', axis=1)
print("📋 DataFrame sin columna 'salario_anual':")
print(df_sin_anual.head())
````

````python
# Eliminar múltiples columnas
df_reducido = df.drop(['salario_anual', 'categoria_edad'], axis=1)
print("📋 DataFrame con columnas eliminadas:")
print(df_reducido.head())
````

````python
# Eliminar filas por índice
df_sin_primera = df.drop(0, axis=0)
print("📋 DataFrame sin la primera fila:")
print(df_sin_primera)
````

## 9️⃣ Valores Nulos (NaN)

````python
# Crear DataFrame con valores nulos para demostrar
data_con_nulos = {
    'nombre': ['Juan', 'María', None, 'Ana', 'Luis'],
    'edad': [25, None, 35, 28, 42],
    'salario': [30000, 45000, None, 35000, 60000]
}
df_nulos = pd.DataFrame(data_con_nulos)

print("📋 DataFrame con valores nulos:")
print(df_nulos)
````

````python
# .isnull() detecta valores nulos (retorna True/False)
print("❓ Detectar valores nulos:")
print(df_nulos.isnull())
````

````python
# .isnull().sum() cuenta valores nulos por columna
print("🔢 Cantidad de valores nulos por columna:")
print(df_nulos.isnull().sum())
````

````python
# .dropna() elimina filas con valores nulos
# how='any': elimina si tiene al menos un nulo
# how='all': elimina solo si todos los valores son nulos

df_sin_nulos = df_nulos.dropna()
print("✅ DataFrame sin filas con valores nulos:")
print(df_sin_nulos)
````

````python
# .fillna() rellena valores nulos
# Puede usar un valor fijo o métodos como 'ffill', 'bfill'

# Rellenar con 0
df_rellenado = df_nulos.fillna(0)
print("0️⃣ Valores nulos rellenados con 0:")
print(df_rellenado)
````

````python
# Rellenar con la media (promedio) de la columna
df_nulos['edad'] = df_nulos['edad'].fillna(df_nulos['edad'].mean())
print("📊 Edad rellenada con promedio:")
print(df_nulos)
````

## 🔟 Agrupación y Agregación

````python
# Volver a crear DataFrame limpio para ejemplos
data = {
    'nombre': ['Juan', 'María', 'Pedro', 'Ana', 'Luis', 'Carmen'],
    'edad': [25, 30, 35, 28, 42, 38],
    'ciudad': ['Madrid', 'Barcelona', 'Madrid', 'Valencia', 'Barcelona', 'Madrid'],
    'salario': [30000, 45000, 55000, 35000, 60000, 48000],
    'departamento': ['Ventas', 'IT', 'IT', 'Ventas', 'Marketing', 'IT']
}
df = pd.DataFrame(data)

print("📋 DataFrame original:")
print(df)
````

````python
# .groupby() agrupa datos por una o más columnas
# Similar a GROUP BY en SQL

# Agrupar por ciudad y calcular promedio de salario
agrupado_ciudad = df.groupby('ciudad')['salario'].mean()
print("💰 Salario promedio por ciudad:")
print(agrupado_ciudad)
````

````python
# Funciones de agregación comunes:
# .sum() - suma
# .mean() - promedio
# .median() - mediana
# .min() - mínimo
# .max() - máximo
# .count() - conteo
# .std() - desviación estándar

# Múltiples agregaciones por departamento
agrupado_dept = df.groupby('departamento')['salario'].agg(['mean', 'min', 'max', 'count'])
print("📊 Estadísticas de salario por departamento:")
print(agrupado_dept)
````

````python
# .agg() con diccionario para diferentes columnas
# Permite aplicar diferentes funciones a diferentes columnas

agrupado_multiple = df.groupby('ciudad').agg({
    'salario': ['mean', 'sum'],  # Promedio y suma de salarios
    'edad': 'mean'  # Promedio de edad
})
print("📊 Múltiples agregaciones por ciudad:")
print(agrupado_multiple)
````

````python
# .value_counts() cuenta valores únicos en una columna
print("🔢 Conteo de personas por ciudad:")
print(df['ciudad'].value_counts())

print("\n" + "="*50 + "\n")

# Con normalize=True muestra porcentajes
print("📊 Porcentaje por ciudad:")
print(df['ciudad'].value_counts(normalize=True) * 100)
````

## 1️⃣1️⃣ Combinar DataFrames

````python
# Crear dos DataFrames para demostrar merge y join
empleados = pd.DataFrame({
    'id': [1, 2, 3, 4],
    'nombre': ['Juan', 'María', 'Pedro', 'Ana'],
    'departamento_id': [10, 20, 10, 30]
})

departamentos = pd.DataFrame({
    'departamento_id': [10, 20, 30],
    'nombre_dept': ['IT', 'Ventas', 'Marketing']
})

print("👥 Empleados:")
print(empleados)
print("\n📊 Departamentos:")
print(departamentos)
````

````python
# .merge() combina DataFrames (similar a JOIN en SQL)
# on: columna común para unir
# how: tipo de join ('inner', 'left', 'right', 'outer')

# Inner join (solo filas que coinciden en ambos)
merged_inner = pd.merge(empleados, departamentos, on='departamento_id', how='inner')
print("🔗 Inner Join:")
print(merged_inner)
````

````python
# Left join (todas las filas del DataFrame izquierdo)
merged_left = pd.merge(empleados, departamentos, on='departamento_id', how='left')
print("⬅️ Left Join:")
print(merged_left)
````

````python
# .concat() apila DataFrames verticalmente u horizontalmente
# axis=0: apilar verticalmente (por defecto)
# axis=1: apilar horizontalmente
# ignore_index=True: crear nuevo índice

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

# Concatenar verticalmente
concat_vertical = pd.concat([df1, df2], ignore_index=True)
print("📚 Concatenación vertical:")
print(concat_vertical)
````

## 1️⃣2️⃣ Pivoting y Reshape

````python
# .pivot_table() crea tabla dinámica (como en Excel)
# values: columna con los valores a agregar
# index: filas de la tabla pivote
# columns: columnas de la tabla pivote
# aggfunc: función de agregación

pivot = df.pivot_table(
    values='salario',
    index='ciudad',
    columns='departamento',
    aggfunc='mean'
)
print("📊 Tabla pivote - Salario promedio por ciudad y departamento:")
print(pivot)
````

## 1️⃣3️⃣ Trabajar con Fechas

````python
# Crear DataFrame con fechas
fechas_data = {
    'fecha': ['2024-01-15', '2024-02-20', '2024-03-10', '2024-04-05'],
    'ventas': [1000, 1500, 1200, 1800]
}
df_fechas = pd.DataFrame(fechas_data)

print("📅 DataFrame con fechas (como texto):")
print(df_fechas)
print(f"\nTipo de 'fecha': {df_fechas['fecha'].dtype}")
````

````python
# pd.to_datetime() convierte texto a fecha
df_fechas['fecha'] = pd.to_datetime(df_fechas['fecha'])

print("📅 Fechas convertidas:")
print(df_fechas)
print(f"\nNuevo tipo de 'fecha': {df_fechas['fecha'].dtype}")
````

````python
# Extraer componentes de fecha
# .dt permite acceder a propiedades de fecha

df_fechas['año'] = df_fechas['fecha'].dt.year
df_fechas['mes'] = df_fechas['fecha'].dt.month
df_fechas['dia'] = df_fechas['fecha'].dt.day
df_fechas['dia_semana'] = df_fechas['fecha'].dt.day_name()  # Nombre del día

print("📅 Componentes de fecha extraídos:")
print(df_fechas)
````

## 1️⃣4️⃣ Aplicar Funciones

````python
# .apply() aplica una función a cada elemento
# Se puede usar con funciones lambda o funciones definidas

# Función para clasificar salarios
def clasificar_salario(salario):
    if salario < 35000:
        return 'Bajo'
    elif salario < 50000:
        return 'Medio'
    else:
        return 'Alto'

# Aplicar función a la columna salario
df['nivel_salario'] = df['salario'].apply(clasificar_salario)
print("💼 Clasificación de salarios:")
print(df[['nombre', 'salario', 'nivel_salario']])
````

````python
# Usar lambda para operaciones simples
# lambda es una función anónima de una sola línea

# Convertir nombres a mayúsculas
df['nombre_mayusculas'] = df['nombre'].apply(lambda x: x.upper())
print("🔤 Nombres en mayúsculas:")
print(df[['nombre', 'nombre_mayusculas']])
````

## 1️⃣5️⃣ Exportar Datos

````python
# .to_csv() guarda DataFrame como CSV
# index=False: no guardar el índice como columna
# encoding='utf-8': usar UTF-8 para caracteres especiales

df.to_csv('datos_procesados.csv', index=False, encoding='utf-8')
print("✅ Archivo 'datos_procesados.csv' guardado exitosamente")
````

````python
# .to_excel() guarda como Excel
# Requiere: pip install openpyxl

df.to_excel('datos_procesados.xlsx', index=False, sheet_name='Empleados')
print("✅ Archivo 'datos_procesados.xlsx' guardado exitosamente")
````

## 🎯 TIPS Y MEJORES PRÁCTICAS

### 1. **Siempre explora tus datos primero**
```python
df.head()
df.info()
df.describe()
df.isnull().sum()
```

### 2. **Usa `inplace=True` con cuidado**
- Modifica el DataFrame original (no se puede deshacer)
- Mejor práctica: crear copias y trabajar con ellas

### 3. **Encadena operaciones para código más limpio**
```python
resultado = (df
    .dropna()
    .sort_values('edad')
    .groupby('ciudad')['salario'].mean()
)
```

### 4. **Verifica tipos de datos después de cargar CSV**
- CSV lee todo como texto por defecto
- Convierte fechas con `pd.to_datetime()`
- Convierte números con `.astype()`

### 5. **Maneja valores nulos apropiadamente**
- No siempre eliminar con `dropna()`
- Considera rellenar con `fillna()` usando media, mediana o moda

### 6. **Usa máscaras booleanas para filtrado complejo**
```python
mask = (df['edad'] > 30) & (df['ciudad'] == 'Madrid')
resultado = df[mask]
```

### 7. **Optimiza memoria con tipos de datos apropiados**
```python
df['categoria'] = df['categoria'].astype('category')  # Ahorra memoria
```

### 8. **Usa `copy()` para evitar SettingWithCopyWarning**
```python
df_nuevo = df[df['edad'] > 30].copy()
```
